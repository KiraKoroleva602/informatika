```
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('iris_data.csv')
a = df['SepalLengthCm']
b = df['SepalWidthCm']
c = df['PetalLengthCm']
d = df['PetalWidthCm']
fig = plt.figure(figsize=(18,12))
ax1 = fig.add_subplot(3,2,1)
ax1.scatter(a,b)
ax2 = fig.add_subplot(3,2,2)
ax2.scatter(a,c)
ax3 = fig.add_subplot(3,2,3)
ax3.scatter(a,d)
ax4 = fig.add_subplot(3,2,4)
ax4.scatter(b,c)
ax5 = fig.add_subplot(3,2,5)
ax5.scatter(b,d)
ax6 = fig.add_subplot(3,2,6)
ax6.scatter(c,d)

mnk = np.polyfit(c,d,1)
z = np.poly1d(mnk)
c1 = np.sort(c)
ax6.plot(c1,z(c1))

mnk = np.polyfit(a,c,1)
z = np.poly1d(mnk)
a1 = np.sort(a)
ax2.plot(a1,z(a1))

mnk = np.polyfit(a,d,1)
z = np.poly1d(mnk)
a1 = np.sort(a)
ax3.plot(a1,z(a1))

plt.tight_layout()
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt


Ax, Ay = 1.0, 1.0
wx, wy = 1.0, 2.0
t = np.linspace(0, 2 * np.pi, 500)
x = Ax * np.sin(wx * t )
y = Ay * np.sin(wy * t )

t1= np.linspace(0, 2 * np.pi, 8, endpoint=False)
x1= Ax * np.sin(wx * t1 )
y1 = Ay * np.sin(wy * t1)

fig = plt.figure(figsize=(4, 12), layout='constrained')

ax_xt = plt.subplot(3,1,1)
ax_xt.plot(t, x, 'g-', label='x(t)')
ax_xt.scatter(t1, x1, color='black', s=20, zorder=5)
ax_xt.set_title('Колебание по оси X')
ax_xt.set_xlabel('Время t')
ax_xt.set_ylabel('X')
ax_xt.grid(True, linestyle=':')

ax_yt = plt.subplot(3,1,2)
ax_yt.plot(t, y, 'b-', label='y(t)')
ax_yt.scatter(t1, y1, color='black', s=20, zorder=5)
ax_yt.set_title('Колебание по оси Y')
ax_yt.set_xlabel('Время t')
ax_yt.set_ylabel('Y')
ax_yt.grid(True, linestyle=':')

ax_l = plt.subplot(3,1,3)
ax_l.plot(x, y, 'r-', linewidth=2, label='Фигура Лиссажу')
ax_l.scatter(x1, y1, color='black', s=20, zorder=5)
ax_l.set_title('Результат сложения: y(x)')
ax_l.set_xlabel('X')
ax_l.set_ylabel('Y')
ax_l.set_xlim(-1.2 * Ax, 1.2 * Ax)
ax_l.set_ylim(-1.2 * Ay, 1.2 * Ay)
ax_l.grid(True, linestyle=':')

plt.show()
```
