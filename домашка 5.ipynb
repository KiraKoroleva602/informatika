```python
from tkinter import *
from random import randint

WIDTH = 300
HEIGHT = 200


class Ball:
    def __init__(self, *args, color = 'green'):
        self.R = randint(10, 50)
        self.x = randint(self.R, WIDTH - self.R)
        self.y = randint(self.R, HEIGHT - self.R)
        self.dx, self.dy = (10, 10)
        self.ball_id = canvas.create_oval(self.x - self.R,
                                     self.y - self.R,
                                     self.x + self.R,
                                     self.y + self.R, fill=color)

    def move(self):
        if self.x + self.R > root.winfo_width() or self.x - self.R <= 0:
            self.dx = -self.dx
        if self.y + self.R > root.winfo_height() or self.y - self.R <= 0:
            self.dy = -self.dy
        self.x += self.dx
        self.y += self.dy

    def show(self):
        canvas.move(self.ball_id, self.dx, self.dy)


def click_handler(event):
    print('Hello World! x=', event.x, 'y=', event.y)


def tick():
    for ball in balls:
        ball.move()
        ball.show()
    root.after(50, tick)


root = Tk()
root.geometry('300x300')
root.update()

canvas = Canvas(root)
canvas.pack(fill=BOTH, expand=True)


new_ball = lambda *args:balls.append(Ball(color = f'#{randint(0,  0xFFFFFF):06x}'))



canvas.bind('<Button-1>',new_ball)
balls = [Ball() for i in range(5)]

tick()
root.mainloop()
```

```python
from tkinter import *
from tkinter import ttk


def calculate(*args):
    try:
        value1 = int(mas.get())
        value2 = float(rost.get())
        value3 = value1/(value2**2/10000)
        r = ''
        if value3 <= 16:
            r = "Выраженный дефицит массы тела"
        elif 16 < value3 <= 18.5:
            r = "Недостаточная (дефицит) масса тела"
        elif 18.5 < value3 <= 25:
            r = "Норма"
        elif 25 < value3 <= 30:
            r = "Избыточная масса тела (предожирение)"
        elif 30 < value3 <= 35:
            r = "Ожирение 1 степени"
        elif 35 < value3 <= 40:
            r = "Ожирение 2 степени"
        else:
            r = "Ожирение 3 степени"

        result.set(f'{value1/(value2**2/10000):.1f} {r}')

    except ValueError:
        pass


root = Tk()
root.title("Calculator")

mainframe = ttk.Frame(root, padding="3 3 12 12")
mainframe.grid(column=0, row=0, sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)
for i in range(10):
    mainframe.columnconfigure(i, weight=1)
    mainframe.rowconfigure(i, weight=1)


mas = StringVar()
mas_entry = ttk.Entry(mainframe, width=7, textvariable= mas)
mas_entry.grid(column=2, row=1, sticky=(W, E))
rost = StringVar()
rost_entry = ttk.Entry(mainframe, width=7, textvariable=rost)
rost_entry.grid(column=2, row=2, sticky=(W, E))


result = StringVar()
ttk.Label(mainframe, textvariable=result).grid(column=2, row=3, sticky=(W, E))

ttk.Button(mainframe, text="Calculate", command=calculate).grid(column=3, row=4, sticky=W)

ttk.Label(mainframe, text="кг").grid(column=3, row=1, sticky=W)
ttk.Label(mainframe, text="см").grid(column=3, row=2, sticky=W)
ttk.Label(mainframe, text="результат").grid(column=3, row=3, sticky=W)

for child in mainframe.winfo_children():
    child.grid_configure(padx=5, pady=5)

mas_entry.focus()
root.bind("<Return>", calculate)

root.mainloop()
```

```python
import random
import tkinter as tk
import pandas as pd


class A:

    def __init__(self, w):
        self.w = w
        self.w.title("Films")
        self.s1 = {}

        d = pd.read_csv("imdb_top_250.csv")
        for i, r in d.iterrows():
            t = r["Title"]
            g_list = r["Genre"].split(" | ")

            for g in g_list:
                g = g.strip()
                if g not in self.s1:
                    self.s1[g] = []
                self.s1[g].append(t)


        self.e = tk.Entry(w)
        self.e.pack()

        self.b = tk.Button(w, text="Find", command=self.find)
        self.b.pack()

        self.l1 = tk.Label(w, text="")
        self.l1.pack()


        genres_string = ", ".join(sorted(self.s1.keys()))
        self.l2 = tk.Label(w, text=genres_string)
        self.l2.pack()

    def find(self):
        g = self.e.get().strip()
        films_list = self.s1[g]
        chosen_film = random.choice(films_list)

        self.l1.config(text=chosen_film + "\n(" + g + ")")


if __name__ == "__main__":
    root = tk.Tk()
    root.geometry("400x250")
    app = A(root)
    root.mainloop()
```

```python
from tkinter import *
from tkinter import ttk


def calculate(*args):
    try:
        color = str(hex1.get()).strip()

        xx_hex = color[1:3]
        yy_hex = color[3:5]
        zz_hex = color[5:7]

        xx = int(xx_hex, 16)
        yy = int(yy_hex, 16)
        zz = int(zz_hex, 16)

        aa = 255 - xx
        bb = 255 - yy
        cc = 255 - zz

        aa_hex = f"{aa:02x}"
        bb_hex = f"{bb:02x}"
        cc_hex = f"{cc:02x}"

        hex2 = "#" + aa_hex + bb_hex + cc_hex

        result.set(str(hex2))

        hex1_entry.config(foreground=color)
        result_label.config(foreground=hex2)

    except:
        pass


root = Tk()
root.title("Calculator")

mainframe = ttk.Frame(root, padding="3 3 12 12")
mainframe.grid(column=0, row=0, sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)
for i in range(10):
    mainframe.columnconfigure(i, weight=1)
    mainframe.rowconfigure(i, weight=1)

hex1 = StringVar()
hex1_entry = ttk.Entry(mainframe, width=7, textvariable=hex1)
hex1_entry.grid(column=2, row=1, sticky=(W, E))

result = StringVar()
result_label = ttk.Label(mainframe, textvariable=result)
result_label.grid(column=2, row=2, sticky=(W, E))

ttk.Button(mainframe, text="Calculate", command=calculate).grid(column=3, row=3, sticky=W)

for child in mainframe.winfo_children():
    child.grid_configure(padx=5, pady=5)

hex1_entry.focus()
root.bind("<Return>", calculate)

root.mainloop()
```

```python
from tkinter import *
from tkinter import ttk
from tkinter import filedialog
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def MNK(x, y):
    a, b = np.polyfit(x, y, 1)
    return a, b

def read_csv(path):
    df = pd.read_csv(path, sep=None, engine="python", decimal=",")
    data = df.to_numpy(dtype=float)
    return data[:, 0], data[:, 1]

def file_select():
    path = filedialog.askopenfilename(
        filetypes=[("CSV", "*.csv"), ("All", "*.*")])
    if path:
        file_path_var.set(path)
        status_var.set("Файл выбран.")

def calculate():
    file_path = file_path_var.get()
    if not file_path:
        status_var.set("Ошибка: выберите CSV-файл.")
        return
    try:
        x, y = read_csv(file_path)
        a, b = MNK(x, y)

        x_line = np.linspace(np.min(x), np.max(x), 200)
        fig = plt.figure(figsize=(8, 5))
        ax = fig.add_subplot(111)
        ax.scatter(x, y, color="tab:blue", label="Экспериментальные точки")
        ax.plot(x_line, a * x_line + b, color="red", linewidth=2,
                label=f"МНК: y = {a:.4f}x + {b:.4f}")
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        ax.set_title("Аппроксимация МНК")
        ax.grid(True, alpha=0.3)
        ax.legend()

        save_path = filedialog.asksaveasfilename(
            defaultextension=".png",
            filetypes=[("PNG", "*.png"), ("PDF", "*.pdf")])
        if save_path:
            fig.savefig(save_path, dpi=200, bbox_inches="tight")
            status_var.set(f"График сохранён: {save_path}")
        else:
            status_var.set("График построен (не сохранён).")

        coef_var.set(f"a = {a:.6f}   b = {b:.6f}")

        plt.show(block=True)
        plt.close(fig)

    except Exception as e:
        status_var.set(f"Ошибка: {e}")

root = Tk()
root.geometry("580x260")

mainframe = ttk.Frame(root, padding="12 12 12 12")
mainframe.grid(column=0, row=0, sticky=(N, W, E, S))
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)

file_path_var = StringVar()
file_path_var.set("файл не выбран")
coef_var = StringVar()
status_var = StringVar()
status_var.set("Готово.")

ttk.Button(mainframe, text="Выбрать файл CSV",
           command=file_select).grid(column=1, row=1,
                                     columnspan=2, sticky=(W, E))

file_label = ttk.Label(mainframe, textvariable=file_path_var,
                       wraplength=500)
file_label.grid(column=1, row=2, columnspan=2, sticky=(W, E))

ttk.Button(mainframe, text="Рассчитать и создать график",
           command=calculate).grid(column=1, row=3,
                                   columnspan=2, sticky=(W, E))

coef_var.set("Коэффициенты МНК: —")
ttk.Label(mainframe, textvariable=coef_var).grid(column=1, row=4,
                                                  columnspan=2, sticky=(W, E))

ttk.Label(mainframe, textvariable=status_var,
          wraplength=500, foreground="gray").grid(column=1, row=5,
                                                   columnspan=2, sticky=(W, E))

for c in mainframe.winfo_children():
    c.grid_configure(padx=7, pady=7)

root.mainloop()
```
