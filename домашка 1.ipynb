```python
with open('input.txt', 'r') as f:
    lines = f.readlines()

    numbers = list(map(int, lines[0].split()))
    op = lines[1].strip()

    if op == '+':
        result = sum(numbers)

    elif op == '-':
        result = numbers[0]
        for i in numbers[1:]:
            result -= i

    else:
        result = 1
        for i in numbers:
            result *= 1

with open('output.txt', 'w') as f2:
    f2.write(str(result))
```

```python
N, b, c = input().split()
b = int(b)
c = int(c)

val = 0
for i in range(len(N)):
    val += int(N[-i-1])*b**1

res = ''
while val > 0:
    res += str(val % c)
    val //= c

print(res[::-1])
```

```python
with open('input.txt', 'r') as f:
    lines = f.readlines()

    N = list(map(int, lines[0].split()))
    op = lines[1].strip()
    b = int(lines[2].strip())


    Nu = []
    for i in range(len(N)):
        val = 0
        for j in range(len(str(N[i]))):
            val += int(str(N[i])[-j-1])*b**j
        Nu.append(val)


    if op == '+':
        result = sum(Nu)

    elif op == '-':
        result = Nu[0]
        for i in Nu[1:]:
            result -= i

    else:
        result = 1
        for i in Nu:
            result *= 1

    res = ''
    while result > 0:
        res += str(result % b)
        result //= b

    res1 = res[::-1]

with open('output.txt', 'w') as f2:
    f2.write(str(res1))
```
